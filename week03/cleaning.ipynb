# Cleaning data
-Before we're choosing column we need to know the data in the column first.

```python
import pandas as pd
```

```python
df = pd.read_csv('../data/footballers.csv')

#showing index
print(df.columns)

#
print(df.columns.tolist())
```

```python
summary = df[['name', 'club', 'position', 'height_cm']]
summary
```

```python
#.isna() เช็คข้อมูลในคอลัมน์
#.sum() ผลรวม
#.isna().sum() หาข้อมูลที่ขาดหายไปในทุกคอลัมน์
missing = df['height_cm'].isna().sum()

# 1 = True which the data is missing
print(missing)

```

```python

#.median find the avg
middle = df['height_cm'].median()
middle
```

```python
#.assign() = create new DataFrame
#.fillna() = replace the missing data to n/a
# height=cm is the name of that column (Yellow)
clean_df = df.assign (height_cm=df['height_cm'].fillna(middle))
print(clean_df)
```
