### Function
- `nlargest`() = เลือก n แถวที่มีค่ามากสุดของคอลัมน์
  
- `plt.bar()` = สร้างกราฟแท่งจากรายชื่อ
   
- `plt.xticks()` = ปรับแต่งข้อความกำกับแกน x 
    
- `plt.ylabel()` = ใส่ชื่อหน่วยให้กับแกนแนวตั้ง
    
- `plt.xlabel()` = ใส่ชื่อหน่วยให้กับแกนแนวนอน

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/footballers.csv")
display(df.head(3))
```

```python
top_5 = df.nlargest(5, "overall")[["name", "overall", "value_millions"]]
display(top_5)
```

```python
condition = 'overall == 95'
Alphonso_Davies_overate = df.query(condition)
Alphonso_Davies_overate["name"]
```

```python

plt.figure(figsize=(8, 4.5))
plt.bar(top_5["name"], top_5["overall"], color='blue')
plt.title('Top 10 Football Players by Overall')
plt.xlabel('Players')
plt.ylabel('Overall')

plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha = 0.7)
plt.savefig("Top_5_Players_by_Overall")
plt.show()
print("Save picture already.")
```
