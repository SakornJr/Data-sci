### Challenge 05 - ประเทศใดมี pace เฉลี่ยสูงสุด?
## ภารกิจ
- บทเรียนใช้ `league` และ `overall` แล้ว - หาค่า `pace` เฉลี่ยแยกตาม `nationality` แล้วเรียงผลจากมากไปน้อย

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/footballers.csv")

```

```python
nationality_pace_mean = df.groupby("nationality")["pace"].mean()
nationality_pace_mean
```

```python
sorted_league = nationality_pace_mean.sort_values(ascending=False).reset_index()
sorted_league
# Define new name column = new name column
sorted_league.columns = ["League", "Average_overall"]

display(sorted_league.round(2))
```
