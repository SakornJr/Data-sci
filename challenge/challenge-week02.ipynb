```python
import pandas as pd
```

```python
data = [
    {"name": "Alice", "department": "CS", "score": 85, "attendance": 95},
    {"name": "Bob", "department": "Math", "score": 78, "attendance": 88},
    {"name": "Carol", "department": "CS", "score": 92, "attendance": 90},
    {"name": "Dora", "department": "Math", "score": 88, "attendance": 91},
    {"name": "Eve", "department": "CS", "score": 76, "attendance": 87},  
]
```

```python
result = pd.DataFrame(data)
result
```

```python

dept_scores = {}


for average in data:
    department = average["department"]
    score = average["score"]
    
    if department not in dept_scores:
        dept_scores[department] = []
    #added department
    
    dept_scores[department].append(score)    
    
dept_avg = {}

for department, scores in dept_scores.items():
    dept_avg[department] = sum(scores) / len(scores)  
    
print("Average score =>", dept_avg)    


    
```
