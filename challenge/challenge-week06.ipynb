```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/footballers.csv")
display(df.head(3))
```

```python
top_5 = df.nlargest(5, "overall")[["name", "overall", "value_millions"]]
display(top_5)
```

```python
plt.figure(figsize=(8, 4.5))
plt.bar(top_5["name"], top_5["overall"], color='blue')
plt.title('Top 5 Football Players by Overall')
plt.xlabel('Players')
plt.ylabel('Overall')

plt.xticks(rotation=15)
plt.grid(axis='y', linestyle='--', alpha = 0.7)
plt.savefig("Top_5_Players_by_Overall")
plt.show()
print("Save picture already.")
```
