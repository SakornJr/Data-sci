#function
- groupby() = แบ่งข้อมูลเป็นกลุ่ม
    df.groupby("group_column")["value_column"]
- mean() = หาค่าเฉลี่ย
    groupby_result.mean() หรือ series.mean()
- agg() = สรุปหลายค่าในครั้งเดียว
    grouped.agg(["count", "mean","median" ])
- sort_values() = เรียงข้อมูล
    series.sort_values(ascending=False)



## OBERVABLE ```Properties``` , ```Function```, ```Method```
- ```Properties``` = don't include () ex ==> .shape, .columns
- ```Methods``` = calling with obj. and inculde () ==> df.query() , isin()

````python
import pandas as pd
````

````python
df = pd.read_csv("../data/footballers.csv")
df.head
````

````python
# New groupby league and find ovr avg
leauge_ovr_mean = df.groupby("league")["overall"].mean()
leauge_ovr_mean
````

````python
filter_leauge = df["league"].isin(["MLS"])
filter_leauge
````

````python
#Sorted values
# ascending=False ... greater to lower
# ascending=True ... lower to greater
# .reset_index() = Not include type index and convert to DataFrame() 
sorted_league = leauge_ovr_mean.sort_values(ascending=False).reset_index()
sorted_league
# Define new name column = new name column
sorted_league.columns = ["League", "Average_overall"]

display(sorted_league.round(2))
````

````python
# .agg()
position_stats = df.groupby("position")["overall"].agg(["count","mean", "median"]).sort_values(by="mean", ascending=False)


position_stats.round(2)
#.agg(["count", "mean", "median"]).sort_values(by="mean", ascending=False)

# posittion_stats.round(2)
````
